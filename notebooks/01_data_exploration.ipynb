## Create Temporary View

Register the source dataset as a temporary SQL view for exploratory analysis.

This view exists only for the current notebook session.

```python
trip_df.createOrReplaceTempView("trip_source")
```

## Basic Fare Profiling

Analyze fare metrics to identify unusual values and potential data quality concerns.

```python
%sql
SELECT
    COUNT(*) AS total_trips,
    AVG(fare_amount) AS avg_fare,
    MAX(fare_amount) AS highest_fare,
    MIN(fare_amount) AS lowest_fare
FROM trip_source
```

## Investigate Negative Fare Records

Review sample records where the fare amount is below zero.

```python
%sql
SELECT
    fare_amount,
    total_amount,
    trip_distance,
    passenger_count
FROM trip_source
WHERE fare_amount < 0
LIMIT 20
```

## Negative Fare Count

Measure the volume of negative fare records in the source data.

```python
%sql
SELECT
    COUNT(*) AS negative_fare_records
FROM trip_source
WHERE fare_amount < 0
```
