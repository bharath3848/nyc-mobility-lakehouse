# GitHub Actions CI/CD

## Objective

Automate Databricks Asset Bundle validation and deployment using GitHub Actions.

### Business Scenario

Manual deployment processes are time-consuming and error-prone.

A CI/CD pipeline enables automated validation and deployment whenever changes are merged into the main branch.

### Components

- GitHub Repository
- GitHub Actions
- Databricks Asset Bundles
- Databricks Workspace

## CI/CD Architecture

Developer
      ↓
Feature Branch
      ↓
Commit
      ↓
Push
      ↓
Pull Request
      ↓
Merge To Main
      ↓
GitHub Actions
      ↓
Bundle Validation
      ↓
Bundle Deployment
      ↓
Databricks Workspace

This workflow automates deployment and reduces manual operational effort.


## GitHub Actions Workflow

The deployment workflow is implemented using:

.github/workflows/databricks-validate.yml

The workflow performs:

1. Checkout Repository
2. Install Databricks CLI
3. Validate Databricks Asset Bundle
4. Deploy Databricks Asset Bundle

Execution occurs automatically when code is merged into the main branch.

## Databricks Asset Bundle Integration

The CI/CD pipeline deploys assets defined in:

- databricks.yml
- resources/workflow.yml

Assets Managed:

- Workflows
- Notebooks
- Configuration
- Pipeline Definitions

Deployment is performed using:

databricks bundle deploy --target dev

## Authentication

GitHub Actions authenticates with Databricks using GitHub Secrets.

Configured Secrets:

### DATABRICKS_HOST

Databricks workspace URL.

### DATABRICKS_TOKEN

Personal Access Token used for automated deployment.

This approach avoids hardcoding credentials inside source code repositories.
