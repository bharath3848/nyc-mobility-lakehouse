## Define JDBC Connection

## Read Customer Data

Load customer records from PostgreSQL into Spark.

```python
customer_df = (
    spark.read
         .format("jdbc")
         .option("url", jdbc_url)
         .option("dbtable", "customers")
         .option("user", "neondb_owner")
         .option("password", "npg_UnJmC6vfAi8K")
         .option("driver", "org.postgresql.Driver")
         .load()
)
```

## Validate Source Data

```python
display(customer_df)
```

```python
customer_df.printSchema()
```

## Add Ingestion Metadata

Capture source and load information for traceability.

```python
from pyspark.sql.functions import (
    current_timestamp,
    lit
)

bronze_customer_df = (
    customer_df
    .withColumn(
        "source_system",
        lit("postgresql")
    )
    .withColumn(
        "ingestion_timestamp",
        current_timestamp()
    )
)
```

```python
display(bronze_customer_df)
```

## Create Bronze Delta Table

```python
bronze_customer_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.bronze_customers"
    )
```
