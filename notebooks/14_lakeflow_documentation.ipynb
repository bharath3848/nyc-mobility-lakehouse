# Lakeflow Declarative Pipelines

## Objective

Implement a managed ETL pipeline using Lakeflow Declarative Pipelines.

### Business Scenario

NYC Taxi trip data arrives through Auto Loader and must be transformed into business-ready datasets using a managed pipeline.

### Source Tables

- workspace.default.bronze_trip_autoloader
- workspace.default.bronze_zones

### Output Tables

- workspace.default.silver_trip_lakeflow
- workspace.default.gold_trip_lakeflow

## Pipeline Architecture

bronze_trip_autoloader
          +
      bronze_zones
          ↓
silver_trip_lakeflow
          ↓
gold_trip_lakeflow

## Pipeline Transformations

### Silver Layer

The Silver layer performs:

- Data quality validation
- Taxi zone enrichment
- Borough derivation
- Business-ready transformation

### Gold Layer

The Gold layer produces:

- Trip volume by pickup borough
- Business reporting metrics

## Data Quality Expectations

The following Lakeflow expectations were implemented:

### valid_trip_distance

trip_distance > 0

### valid_fare_amount

fare_amount >= 0

### valid_passenger_count

passenger_count > 0

Records violating these rules are automatically dropped by the pipeline.
