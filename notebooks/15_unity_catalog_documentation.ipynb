# Unity Catalog

## Objective

Explore Unity Catalog and understand how Databricks manages and governs data assets.

### Why Unity Catalog?

Unity Catalog provides centralized governance for:

- Catalogs
- Schemas
- Tables
- Volumes
- Lineage
- Permissions
- Metadata Discovery

## Information Schema

Information Schema contains metadata about catalogs, schemas and tables.

It supports metadata discovery and governance reporting.

```python
%sql
SELECT
    table_catalog,
    table_schema,
    table_name,
    table_type
FROM workspace.information_schema.tables
ORDER BY table_name;
```

## Lineage

Unity Catalog automatically tracks data lineage.

Project Example

bronze_trip_autoloader
          +
      bronze_zones
          ↓
silver_trip_lakeflow
          ↓
gold_trip_lakeflow

Benefits:

- Dependency tracking
- Impact analysis
- Root cause analysis
- Data discovery

## Permissions

Unity Catalog provides centralized access control.

Permissions can be managed at:

- Catalog level
- Schema level
- Table level
- Volume level

This supports secure and governed access to data assets.

## Governance Features

Unity Catalog centralizes:

- Metadata Management
- Data Discovery
- Access Control
- Lineage Tracking
- Storage Governance

These capabilities improve platform governance and operational visibility.

## Search And Discovery

Unity Catalog enables users to:

- Search datasets
- Explore metadata
- Discover data assets
- View lineage
- Navigate dependencies

This improves self-service analytics and data discoverability.
