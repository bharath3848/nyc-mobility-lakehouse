# Silver Layer - Weather Data

## Objective

Standardize weather data for analytical use.

### Input

- workspace.default.bronze_weather

### Transformations

- Convert weather_date to date datatype
- Validate weather metrics
- Create Silver weather table

### Output

- workspace.default.silver_weather

## Read Bronze Weather Table

```python
from pyspark.sql.functions import (
    col,
    to_date
)

bronze_weather = spark.table(
    "workspace.default.bronze_weather"
)
```

```python
bronze_weather.printSchema()
```

## Standardize Date Column

Convert weather_date from string to date datatype.

```python
silver_weather_df = bronze_weather.withColumn(
    "weather_date",
    to_date(col("weather_date"))
)
```

## Validate Updated Schema

```python
silver_weather_df.printSchema()
```

```python
display(silver_weather_df)
```

## Data Quality Review

Review weather metrics before publishing the Silver dataset.

```python
silver_weather_df.createOrReplaceTempView(
    "silver_weather_view"
)
```

```python
%sql
SELECT
    MIN(max_temperature) AS min_max_temp,
    MAX(max_temperature) AS max_max_temp,
    MIN(min_temperature) AS min_min_temp,
    MAX(min_temperature) AS max_min_temp,
    MAX(precipitation) AS highest_precipitation
FROM silver_weather_view
```

## Create Silver Weather Table

```python
silver_weather_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.silver_weather"
    )
```

```python
%sql
SELECT COUNT(*)
FROM workspace.default.silver_weather
```

```python
%sql
SELECT *
FROM workspace.default.silver_weather
ORDER BY weather_date
```
